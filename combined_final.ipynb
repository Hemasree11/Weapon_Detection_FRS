```python
import cv2
import numpy as np
import mediapipe as mp
from tensorflow.keras.models import load_model
import torch
from ultralytics import YOLO

# Load models
custom_person_model = load_model('facerecognition_finalmodel.h5')  # Custom CNN for person detection
yolo_weapon_model = YOLO(r"D:\wepon_detection\runs\detect\train12\weights\best.pt")  # YOLO for weapon detection
face_name = np.load('class_indices.npy', allow_pickle=True).item()
# Mediapipe pose setup
mp_pose = mp.solutions.pose
pose_estimator = mp_pose.Pose(static_image_mode=True, model_complexity=2)
def preprocess_for_cnn(image):
    """
    Detect faces in the image using the face detection model.
    """
    face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    faces = face_cascade.detectMultiScale(gray_image, scaleFactor=1.3, minNeighbors=5, minSize=(30, 30))
    face_detections = []

    for (x, y, w, h) in faces:
        face_roi = image[y:y+h, x:x+w]
        face_resized = cv2.resize(face_roi, (100, 100))
        face_resized = cv2.cvtColor(face_resized, cv2.COLOR_BGR2RGB)
        face_input = np.expand_dims(face_resized, axis=0) / 255.0
    return face_input
def decode_predictions_to_boxes(predictions, class_name="person", confidence_threshold=0.5):
    """
    Decode predictions from the custom CNN into bounding boxes.
    :param predictions: Model predictions containing bounding boxes and confidence scores.
    :param class_name: Class to filter for (e.g., 'person').
    :param confidence_threshold: Minimum confidence score to consider a detection.
    :return: List of bounding boxes for the specified class.
    """
    boxes = []
    for prediction in predictions:
        # Assume each prediction is a dictionary with 'box', 'class', and 'confidence'
        if prediction["class"] == class_name and prediction["confidence"] >= confidence_threshold:
            boxes.append(prediction["box"])
    return boxes
 
# Function to detect persons using the custom CNN model
def detect_persons_with_custom_cnn(model, image):
    """
    Detect persons using the custom CNN model.
    :param model: Trained CNN model for person detection.
    :param image: Input image.
    :return: List of bounding boxes for detected persons.
    """
    input_image = preprocess_for_cnn(image)  # Preprocess image for your CNN
    predictions = model.predict(input_image)
    print(len(predictions))# Decode CNN output
    print("Predictions:", predictions)
    print("Shape of predictions:", np.array(predictions).shape)

    class_name = np.load('class_indices.npy', allow_pickle=True).item()
    person_boxes = decode_predictions_to_boxes(predictions, class_name)
    
    return person_boxes

# Function to estimate pose using Mediapipe
def estimate_pose(image, person_box):
    """
    Estimate pose using Mediapipe.
    :param image: Input image.
    :param person_box: Bounding box of the person.
    :return: Keypoints for the detected person.
    """
    cropped_person = crop_image(image, person_box)
    cropped_person_rgb = cv2.cvtColor(cropped_person, cv2.COLOR_BGR2RGB)
    result = pose_estimator.process(cropped_person_rgb)

    keypoints = {}
    if result.pose_landmarks:
        for idx, landmark in enumerate(result.pose_landmarks.landmark):
            keypoints[f"point_{idx}"] = (
                landmark.x * cropped_person.shape[1],
                landmark.y * cropped_person.shape[0],
            )
    return keypoints
def detect_objects(model, image, class_name=None, confidence_threshold=0.5):
    """
    Detect objects using the YOLO model.
    :param model: YOLO model.
    :param image: Input image.
    :param class_name: Class name to filter detections.
    :param confidence_threshold: Minimum confidence score.
    :return: List of bounding boxes.
    """
    results = model(image)  # Get detections
    boxes = []
    for result in results.xyxy[0].cpu().numpy():  # Adjust indexing based on YOLO's output
        x_min, y_min, x_max, y_max, confidence, class_id = result
        if confidence >= confidence_threshold and model.names[int(class_id)] == class_name:
            boxes.append([int(x_min), int(y_min), int(x_max), int(y_max)])
    return boxes


def decode_predictions_to_boxes(predictions, class_name, confidence_threshold=0.5):
    """
    Decode model predictions into bounding boxes for the specified class.
    """
    # Example structure: [x_min, y_min, x_max, y_max, confidence, class_id]
    class_map = {1: "person"}  # Adjust based on your model
    boxes = []
    
    for pred in predictions:
        x_min, y_min, x_max, y_max, confidence, class_id = pred
        if confidence >= confidence_threshold and class_map.get(class_id) == class_name:
            boxes.append([int(x_min), int(y_min), int(x_max), int(y_max)])
    
    return boxes


def crop_image(image, box):
    """
    Crop the image to the specified bounding box.
    """
    x_min, y_min, x_max, y_max = box
    return image[y_min:y_max, x_min:x_max]


def annotate_image(image, box, label):
    """
    Annotate the image with a bounding box and label.
    """
    x_min, y_min, x_max, y_max = box
    color = (0, 255, 0)  # Green for bounding boxes
    cv2.rectangle(image, (x_min, y_min), (x_max, y_max), color, 2)
    cv2.putText(image, label, (x_min, y_min - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, color, 2)

# Step 1: Detect persons and weapons
image = cv2.imread('single_with_weapon.jpeg')
person_boxes = detect_persons_with_custom_cnn(custom_person_model, image)
weapon_boxes = detect_objects(yolo_weapon_model, image, class_name='weapon')

# Step 2: Get keypoints for each detected person
persons_with_keypoints = []
for person_box in person_boxes:
    keypoints = estimate_pose(image, person_box)
    persons_with_keypoints.append({"box": person_box, "keypoints": keypoints})

# Step 3: Match weapons to persons
weapon_to_person_mapping = []
for weapon_box in weapon_boxes:
    weapon_center, weapon_orientation = compute_weapon_direction(weapon_box)
    closest_person = None
    min_distance = float('inf')

    for person in persons_with_keypoints:
        keypoints = person["keypoints"]
        closest_hand = find_closest_hand(weapon_center, keypoints)
        
        if closest_hand:
            arm_joint = keypoints.get('point_11') or keypoints.get('point_12')  # Example indices for elbow joints
            if arm_joint and is_weapon_direction_valid(weapon_orientation, closest_hand, arm_joint):
                distance = np.linalg.norm(np.array(weapon_center) - np.array(closest_hand))
                if distance < min_distance:
                    min_distance = distance
                    closest_person = person

    if closest_person:
        weapon_to_person_mapping.append({"weapon": weapon_box, "person": closest_person})

# Step 4: Annotate results
for mapping in weapon_to_person_mapping:
    person_box = mapping["person"]["box"]
    weapon_box = mapping["weapon"]
    annotate_image(image, person_box, "Holding Weapon")
    annotate_image(image, weapon_box, "Weapon")

# Save or display the output
cv2.imshow('Result', image)
cv2.imwrite('output_image.jpg', image)
```
